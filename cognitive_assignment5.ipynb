Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:
a) Addition of 2 in all the element  
b) Multiply 3 with all the elements
c) Divide every element by 2

```python
import numpy as np
a=np.array([1,2,3,4,5])
print("Original array: ",a)
print("after addition of 2: ",a+2)
print("after multiply with 3: ",a*3)
print("after division by 2: ",a/2)
```

Q.2 Questions on Basic NumPy Array: a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])

```python
arr=np.array([1,2,3,6,4,5])
print("original: ",arr)
print("reversed: ",np.flip(arr))
```

b) Find the most frequent value and their indice(s) in the following arrays: i. x = np.array([1,2,3,4,5,1,2,1,1,1])

```python
x=np.array([1,2,3,4,5,1,2,1,1,1])
value,count=np.unique(x,return_counts=True)
max_count=count.max()
most_freq=value[count==max_count]
indices=np.where(x==most_freq)[0]
print("most frequent value: ",most_freq)
print("frequency: ",max_count)
print("indices: ",indices)
```

ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
y=np.array([1,1,1,2,3,4,2,4,3,3])
values,counts=np.unique(y,return_counts=True)
max_count=counts.max()
most_frequent=values[counts==max_count]
indices=np.where(np.isin(y,most_frequent))[0]
print("most frequent value: ",most_frequent)
print("frequency: ",max_count)
print("indices: ",indices)
```

Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?

```python
amisha=np.linspace(10,100,25)
print("array: ",amisha)
print("dimension of array: ",amisha.ndim)
print("shape of array: ",amisha.shape)
print("total elements: ",amisha.size)
print("data type of each element: ",amisha.dtype)
print("total no. of bytes used by array: ",amisha.nbytes)

transpose=amisha.reshape(25,1)
print(transpose)
t=amisha.reshape(1,25).T
print(t)
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_<your_name>> with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ ucs420_<your_name>>. Resize the array to two rows and three columns and name it as resized_ ucs420_<your_name>>.

```python
ucs420_amisha=np.array([
    [10,20,30,40],
    [50,60,70,80],
    [90,15,20,35]
])
print("mean: ",np.mean(ucs420_amisha))
print("median : ",np.median(ucs420_amisha))
print("max: ",np.max(ucs420_amisha))
print("min: ",np.min(ucs420_amisha))
print("unique values: ",np.unique(ucs420_amisha))
reshaped_ucs420_amisha=ucs420_amisha.reshape(4,3)
print("reshaped: ",reshaped_ucs420_amisha)
resized_ucs420_amisha=np.resize(ucs420_amisha,(2,3))
print("resized: ",resized_ucs420_amisha)
```
